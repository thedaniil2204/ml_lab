```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
true_b = np.random.uniform(-3, 3)
true_k = np.random.uniform(-3, 3)

X = np.linspace(0,10, 100)
Y = true_b + true_k * X - 0.5*np.random.randn(*X.shape)
```

```python
def train_step(Ypoints, Xpoints, border, lr=0.1):
    output = (Ypoints - (border[0] + border[1]*Xpoints))**2
    
    loss_old = output
    loss_old = np.sum(loss_old)

    delta = np.random.normal(size=(2))
    new_border = border + lr*delta

    output = (Ypoints - (new_border[0] + new_border[1]*Xpoints))**2

    loss_new = output
    loss_new = np.sum(loss_new)

    if loss_old > loss_new:
        border = new_border

    return border
```

```python
border = np.random.normal(size=(2))
history = []
for _ in range(500):
    border = train_step(Y, X, border, lr=0.1)
    history.append([border[0],border[1]])
```

```python
Y_true = border[0] + border[1]*X

tolerance = 1
correct = np.abs(Y_true - Y) <= tolerance
accuracy = np.mean(correct)
print(f"Accuracy: {accuracy:.2%}")

plt.scatter(X,Y)
plt.plot(X,Y_true, '--', color='red')
```

```python
history = np.array(history)

plt.plot(history[:,0], history[:,1])  # соединяем путь линией
plt.scatter(history[0,0], history[0,1], color="red", s=120, marker="x", label="Старт", zorder=5)
plt.scatter(history[-1,0], history[-1,1], color="blue", s=120, marker="*", label="Финиш", zorder=5)
plt.scatter(true_b, true_k, color="green", s=120, marker=".", label="Истина", zorder=5)

plt.grid()
plt.xlabel("Вес b")
plt.ylabel("Вес k")
plt.title("Путь весов (b, k) в процессе обучения")
plt.legend()
```
