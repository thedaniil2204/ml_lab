```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class NeuralNetBiasReg:
    def __init__(self, hidden_size=32, seed=42):
        rng = np.random.default_rng(seed)
        # Xavier-подобная инициализация для скрытого слоя
        self.w1 = rng.normal(0, np.sqrt(1 / 1), size=(hidden_size, 1))
        self.bias1 = np.zeros((hidden_size, 1))
        # Малые выходные веса помогают избежать слишком больших начальных ответов
        self.w2 = rng.normal(0, 0.1, size=(1, hidden_size))
        self.bias2 = np.zeros((1, 1))

    @staticmethod
    def activation(x):
        # Ограничение аргумента предотвращает переполнение exp на больших значениях
        x = np.clip(x, -60, 60)
        return 1.0 / (1.0 + np.exp(-x))

    def forward(self, x):
        self.hidden = self.activation(self.w1 @ x + self.bias1)
        self.output = self.w2 @ self.hidden + self.bias2  # линейный выход
        return self.output

    @staticmethod
    def loss(output, target):
        return np.mean((output - target) ** 2)

    def train_step(self, x, target, lr=0.05):
        n = x.shape[1]
        prediction = self.forward(x)

        # d(MSE)/d(output)
        d_output = 2 * (prediction - target) / n
        d_w2 = d_output @ self.hidden.T
        d_bias2 = np.sum(d_output, axis=1, keepdims=True)

        # Обратный проход через линейный выход и сигмоиду скрытого слоя
        d_hidden = self.w2.T @ d_output
        d_hidden_pre = d_hidden * self.hidden * (1 - self.hidden)
        d_w1 = d_hidden_pre @ x.T
        d_bias1 = np.sum(d_hidden_pre, axis=1, keepdims=True)

        self.w2 -= lr * d_w2
        self.bias2 -= lr * d_bias2
        self.w1 -= lr * d_w1
        self.bias1 -= lr * d_bias1
        return self.loss(prediction, target)
```

```python
rng = np.random.default_rng(7)
X = np.linspace(-5, 5, 200)
Y_true = np.sin(X) + 0.5 * X
Y = Y_true + 0.1 * rng.normal(size=X.shape)

X_train = X[np.newaxis, :]
Y_train = Y[np.newaxis, :]

reg_model = NeuralNetBiasReg(hidden_size=32, seed=42)
for _ in range(10_000):
    reg_model.train_step(X_train, Y_train, lr=0.05)

output = reg_model.forward(X_train)
plt.scatter(X, output[0], s=16, label="Предсказание модели")
plt.scatter(X, Y, s=10, alpha=0.35, label="Обучающие точки с шумом")
plt.plot(X, Y_true, color="orange", label="Исходная функция")
plt.grid()
plt.legend()
plt.show()
```

```python
def regression_metrics(y_true, y_pred):
    error = y_pred - y_true
    mae = np.mean(np.abs(error))
    rmse = np.sqrt(np.mean(error ** 2))
    ss_res = np.sum(error ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    r2 = 1 - ss_res / ss_tot
    return mae, rmse, r2

mae, rmse, r2 = regression_metrics(Y_true, output[0])
print(f"MAE={mae:.4f}, RMSE={rmse:.4f}, R²={r2:.4f}")
```
